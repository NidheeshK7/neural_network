```python
import pandas as pd
```

```python
df_train=pd.read_csv('train.csv')
df_train.head()
```

```python
df_test=pd.read_csv('test.csv')
df_test.head()
```

```python
print(df_train.shape)
print(df_test.shape)
```

```python
X_train  = df_train.iloc[:, 1:].values
y_train = df_train.iloc[:, 0].values
X_test=df_test
```

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

```python
import numpy as np
from NN_Numpy import LinearLayer, ReLU, CrossEntropyLoss, SGD, Model

num_classes = 10
y_train_onehot = np.eye(num_classes)[y_train]

model = Model()
model.add_layer(LinearLayer(784, 128))
model.add_layer(ReLU())
model.add_layer(LinearLayer(128, 64))
model.add_layer(ReLU())
model.add_layer(LinearLayer(64, 10))

loss_fn = CrossEntropyLoss()
optimizer = SGD(lr=0.1) 
model.compile(loss_fn, optimizer)

model.train(X_train, y_train_onehot, epochs=50, batch_size=32)

print("\nEvaluation on Training Data:")
train_loss, train_acc = model.evaluate(X_train, y_train_onehot)
print(f"Final Training Accuracy: {train_acc:.2%}")
```

```python
if hasattr(X_test, 'values'):
    X_test_np = X_test.values
else:
    X_test_np = X_test

logits = model.predict(X_test_np)

predicted_labels = np.argmax(logits, axis=1)

submission = pd.DataFrame({
    'ImageId': np.arange(1, len(predicted_labels) + 1),
    'Label': predicted_labels
})

submission.to_csv('submission.csv', index=False)
print("'submission.csv' saved")
print(submission.head())
```
